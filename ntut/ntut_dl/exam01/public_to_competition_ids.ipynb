# Convert public-dataset test predictions to competition IDs

If you trained on the public **AR Traffic Objects** dataset, your test predictions are keyed by the public filenames (`00000001.jpg`, `fence_01.jpg`, …). The competition scores the same 249 images under the IDs `test_0001` … `test_0249`, in a different order. This notebook converts your predictions to those IDs.

The images are the same size in both versions, so your normalized boxes (`class conf cx cy w h`) stay unchanged. Only the `image_id` changes.

Use **one** of the two options below:

- **Option A:** you already have a CSV with one row per public test image (`image_id`, `PredictionString`).
- **Option B:** you have trained weights (`best.pt`); the notebook predicts the public test images and writes the submission.

**Reminder:** the public dataset also contains the test labels (`test/real/labels`). Using them in any way is not allowed.

```python
from pathlib import Path
import pandas as pd


def find_public_test_dir():
    # The public dataset's test images: .../test/real/images under /kaggle/input.
    hits = sorted(Path("/kaggle/input").glob("**/test/real/images"))
    return hits[0] if hits else None


def find_sample_submission():
    hits = sorted(Path("/kaggle/input").glob("**/sample_submission.csv"))
    return hits[0] if hits else None


PUBLIC_TEST_DIR = find_public_test_dir()
SAMPLE_SUB = find_sample_submission()
WORK = Path("/kaggle/working") if Path("/kaggle/working").exists() else Path(".").resolve()
print("public test images:", PUBLIC_TEST_DIR)
print("sample submission :", SAMPLE_SUB)


IMG_EXTS = {".jpg", ".jpeg", ".png"}


def public_stem(name):
    # "test/real/images/a_jpg.rf.123.jpg" -> "a_jpg.rf.123"; names without an image extension stay as they are
    p = Path(str(name))
    return p.stem if p.suffix.lower() in IMG_EXTS else p.name


def to_competition_ids(df):
    # df: image_id = public filename (with or without extension or folder), PredictionString.
    df = df.copy()
    stems = df["image_id"].map(public_stem)
    unknown = sorted(set(stems) - set(PUBLIC_TO_ID))
    if unknown:
        raise ValueError(f"{len(unknown)} ids are not public test filenames, e.g. {unknown[:3]}")
    df["image_id"] = stems.map(PUBLIC_TO_ID)
    if df["image_id"].duplicated().any():
        raise ValueError("the same test image appears more than once")
    missing = sorted(set(PUBLIC_TO_ID.values()) - set(df["image_id"]))
    if missing:
        print(f"warning: {len(missing)} test images have no prediction; writing 'none' for them")
        df = pd.concat([df, pd.DataFrame({"image_id": missing, "PredictionString": "none"})])
    df["PredictionString"] = df["PredictionString"].fillna("none").astype(str)
    df.loc[df["PredictionString"].str.strip() == "", "PredictionString"] = "none"
    order = (pd.read_csv(SAMPLE_SUB)["image_id"].tolist() if SAMPLE_SUB is not None
             else sorted(PUBLIC_TO_ID.values()))
    return df.set_index("image_id").loc[order].reset_index()[["image_id", "PredictionString"]]
```

## 2. Option A: convert an existing prediction CSV

Set `MY_PREDICTIONS` to your file. Its `image_id` column must hold the **public** filenames, for example `00000001` or `fence_01.jpg`.

```python
MY_PREDICTIONS = None   # e.g. "/kaggle/working/my_public_predictions.csv"

if MY_PREDICTIONS:
    submission = to_competition_ids(pd.read_csv(MY_PREDICTIONS, keep_default_na=False))
    submission.to_csv(WORK / "submission.csv", index=False)
    print("saved", WORK / "submission.csv", len(submission), "rows")
    display(submission.head())
```

## 3. Option B: predict the public test images with your weights

Set `WEIGHTS` to your trained model and `IMGSZ` to the size you trained with.

```python
WEIGHTS = None          # e.g. "/kaggle/working/runs/yolo11n/weights/best.pt"
IMGSZ = 640

if WEIGHTS:
    import torch
    from ultralytics import YOLO

    assert PUBLIC_TEST_DIR is not None, "add the public AR Traffic Objects dataset to this notebook"
    model = YOLO(WEIGHTS)
    files = sorted(p for p in PUBLIC_TEST_DIR.iterdir() if p.suffix.lower() in IMG_EXTS)
    rows = []
    for i in range(0, len(files), 16):
        batch = [str(f) for f in files[i:i + 16]]
        results = model.predict(batch, imgsz=IMGSZ, conf=0.001, max_det=100,
                                device=0 if torch.cuda.is_available() else "cpu", verbose=False)
        for f, r in zip(batch, results):  # results come back in input order
            b = r.boxes
            s = " ".join(f"{int(c)} {p:.4f} {x:.6f} {y:.6f} {w:.6f} {h:.6f}"
                         for c, p, (x, y, w, h) in zip(b.cls.tolist(), b.conf.tolist(), b.xywhn.tolist()))
            rows.append((Path(f).stem, s or "none"))
    submission = to_competition_ids(pd.DataFrame(rows, columns=["image_id", "PredictionString"]))
    submission.to_csv(WORK / "submission.csv", index=False)
    print("saved", WORK / "submission.csv", len(submission), "rows")
    display(submission.head())
```
